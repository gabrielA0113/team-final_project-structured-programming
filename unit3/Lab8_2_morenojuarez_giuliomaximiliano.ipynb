# Part 2. What Is a Procedure?

Example:



````python
def calculate_area(base, height):
    return base * height

area = calculate_area(5, 4)

print(area)
````

````python
def display_menu():
    print("1. Add")
    print("2. Subtract")
    print("3. Exit")

display_menu()
````

A recursive function is a function that calls itself.

Observe the following example.



````python
def countdown(n):

    print(n)

    if n > 1:
        countdown(n - 1)

countdown(5)
````



Execute the program.

### Questions

1. Which numbers are displayed?

## The numbers 5, 4, 3, 2, 1 are displayed

2. What happens when `n` becomes 1?

## The condition if n > 1: becomes false. Therefore, the function skips the recursive call and finishes executing without calling itself again.
---

3. Why does the program stop?

## It stops because it reaches the condition where n is no longer greater than 1, preventing any further recursive calls.
---



# Part 7. Understanding the Base Case

Every recursive function needs a condition that stops the recursion.

Observe the following example.



````python
def say_hello(n):

    if n == 0:
        return

    print("Hello")
    say_hello(n - 1)

say_hello(4)
````

### Questions

1. How many times is "Hello" printed?

## It is printed 4 times.
---

2. What is the base case?

## The base case is if n == 0:, which terminates the recursion by executing an empty return.

---

3. What would happen if the `if` statement were removed?

## The function would keep calling itself infinitely (say_hello(-1), say_hello(-2), etc.) until Python runs out of memory and crashes with an error: maximum recursion depth exceeded.
---


# Part 8. Building a Recursive Function

Complete the following function.



````python
def count_to_one(n):

    print(n)

    if n > 1:
        count_to_one(n - 1)

count_to_one(10)
````

Observe carefully how the function calls itself.

### Reflection

What is the role of the statement:

```python
count_to_one(n - 1)
```

*Answer*

This is the recursive call. It instructs the function to call a new instance of itself with a decremented value of **n** (one less than the current value), driving the sequence down toward the base case.

# Part 9. First Recursive Mathematical Example

The factorial of a number is defined as:

```text
4! = 4 × 3 × 2 × 1
```

Observe the following code.

```python
def factorial(n):

    if n == 1:
        return 1

    return n * factorial(n - 1)

print(factorial(4))
```

### Activity

Complete the table.

| factorial(3)  |        3 * factorial(2)     |
| factorial(2)  |      2 * factorial(1)       |
| factorial(1)  |      Returns 1 directly       |

---

### Reflection

Why does the function eventually stop?

# The function stops because it eventually hits the base case if n == 1:, which immediately returns 1 without making another recursive call, allowing all the pending multiplications to resolve.

````python
#Part 10. Final Challenge (CONTAINS ADDITIONAL PRACTICE REQUIREMENTS)

# PROCEDURES

def display_menu():
    print("\n" + "="*25)
    print("1. Calculate Rectangle/Parallelogram Area")
    print("2. Calculate Average of Three Numbers")
    print("3. Recursive Countdown")
    print("4. Check if Number is Even or Odd")  # Menu Option
    print("5. Exit")
    print("="*25)

def say_goodbye():
    print("\nThank you for using the program. Goodbye!")


# FUNCTIONS WITH ROBUST VALIDATION

def get_valid_float(prompt, allow_negative=False):
    """Handles both error handling (type crashing) and input validation (logical values)."""
    while True:
        try:
            value = float(input(prompt))
            if not allow_negative and value <= 0:
                print("Error: The value must be greater than zero. Please try again.")
                continue
            return value
        except ValueError:
            print("Invalid input! Please enter a valid numerical value.")

def get_valid_int(prompt, minimum_val=None):
    """Validates integer inputs for options like countdown."""
    while True:
        try:
            value = int(input(prompt))
            if minimum_val is not None and value < minimum_val:
                print(f"Error: The value must be at least {minimum_val}.")
                continue
            return value
        except ValueError:
            print("Invalid input! Please enter a whole number.")

def calculate_area(base, height):
    return base * height

def calculate_average(num1, num2, num3):
    return (num1 + num2 + num3) / 3


# RECURSIVE FUNCTION

def countdown(n):
    print(n)
    if n > 1:
        countdown(n - 1)


# PROGRAM LOOP

def main():
    while True:
        display_menu()
        choice = input("Select an option (1-5): ").strip()

        if choice == '1':
            print("\n--- Calculate Area ---")
            b = get_valid_float("Enter the base length: ")
            h = get_valid_float("Enter the height: ")
            area = calculate_area(b, h)
            print(f"Success: A shape with a base of {b} and height of {h} has a total area of {area} square units.")

        elif choice == '2':
            print("\n--- Calculate Average ---")
            # Allows negative numbers since an average can be negative
            n1 = get_valid_float("Enter first number: ", allow_negative=True)
            n2 = get_valid_float("Enter second number: ", allow_negative=True)
            n3 = get_valid_float("Enter third number: ", allow_negative=True)
            avg = calculate_average(n1, n2, n3)
            print(f"Result: The mathematical average of [{n1}, {n2}, {n3}] is {avg:.2f}.")

        elif choice == '3':
            print("\n--- Countdown ---")
            start = get_valid_int("Enter a positive starting integer: ", minimum_val=1)
            print(f"Starting recursive countdown from {start}:")
            countdown(start)

        elif choice == '4':  # Additional menu option
            print("\n--- Even or Odd Checker ---")
            num = get_valid_int("Enter any integer to check: ")
            if num % 2 == 0:
                print(f"The number {num} is Even.")
            else:
                print(f"The number {num} is Odd.")

        elif choice == '5':
            say_goodbye()
            break

        else:
            # Error handling for invalid choice
            print("⚠️ Selection Error: 'Let's try that again.' Please choose a valid option from 1 to 5.")

main()
````
